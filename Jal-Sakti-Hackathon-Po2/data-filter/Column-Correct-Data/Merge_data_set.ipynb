```python
import pandas as pd
import os

# Files ki list (Jo tumne upload ki thi)
files = [
    "new_2012-2013.csv", "new_2016-2017.csv", "new_2019-2020.csv",
    "new_2021-2022.csv", "new_2022-2023.csv", "new_2023-2024.csv",
    "new_2024-2025.csv"
]

all_data = []

for file in files:
    # CSV read karo
    df = pd.read_csv(file)
    
    # Filename se Year nikalo (e.g., 'new_2012-2013.csv' -> '2012-2013')
    year_label = file.replace("new_", "").replace(".csv", "")
    
    # Ek naya column 'Year' add karo
    df['Year'] = year_label
    
    # Column names ko clean karo (extra spaces hatao)
    df.columns = df.columns.str.strip()
    
    all_data.append(df)

# Saari files ko jod do (Merge)
master_df = pd.concat(all_data, ignore_index=True)

# Final file save karo
master_df.to_csv("ingres_master_data.csv", index=False)

print("Data merge successful! File saved as 'ingres_master_data.csv'")
print(master_df.head())
```
