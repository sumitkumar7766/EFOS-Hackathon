```python
df_raw = pd.read_excel("./Pre-Data/2012-2013.xlsx", header=None)
```

```python
header_rows = df_raw.iloc[7:10].copy()
```

```python
header_rows[8]
```

```python
header_rows = header_rows.ffill(axis=1)
header_rows
```

```python
new_columns = []
for i in range(header_rows.shape[1]):
    # Har column ke liye charo rows ki values uthayein
    parts = header_rows.iloc[:, i].astype(str).tolist()
    
    # 'nan' aur extra spaces hatayein
    clean_parts = [p.strip() for p in parts if p.lower() != 'nan' and p.strip() != '']
    
    # Inhe '_' se jod de (Example: "Rainfall_Monsoon_Total")
    col_name = "_".join(clean_parts)
    new_columns.append(col_name)
```

```python
df_final = pd.read_excel("./Pre-Data/2012-2013.xlsx", skiprows=11, header=None)
```

```python
df_final.columns = new_columns
```

```python
if df_final.iloc[0,0] == 'S.No': 
    df_final = df_final.iloc[1:].reset_index(drop=True)
```

```python
print("New Columns Sample:", df_final.columns[:5])
print(df_final.head())
```

```python
df_final.to_csv('new_2012-2013.csv', index=False)
```
